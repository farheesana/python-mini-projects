```python
# ==========================================
#       BANK MANAGEMENT SYSTEM
# ==========================================

print("===================================")
print("       BANK MANAGEMENT SYSTEM")
print("===================================")

# Account details
account_name = ""
account_number = ""
balance = 0.0
account_created = False


# Function to create an account
def create_account():
    global account_name, account_number, balance, account_created

    if account_created:
        print("\nAn account already exists.")
        return

    print("\n----- CREATE ACCOUNT -----")

    account_name = input("Enter your name: ")
    account_number = input("Enter account number: ")

    initial_deposit = float(input("Enter initial deposit: "))

    if initial_deposit < 0:
        print("Deposit cannot be negative.")
        return

    balance = initial_deposit
    account_created = True

    print("\nAccount created successfully!")
    print("Account Holder:", account_name)
    print("Account Number:", account_number)
    print("Balance: ₹", balance)


# Function to deposit money
def deposit_money():
    global balance

    if not account_created:
        print("\nPlease create an account first.")
        return

    amount = float(input("\nEnter amount to deposit: "))

    if amount <= 0:
        print("Please enter a valid amount.")
    else:
        balance = balance + amount
        print("Amount deposited successfully!")
        print("Current balance: ₹", balance)


# Function to withdraw money
def withdraw_money():
    global balance

    if not account_created:
        print("\nPlease create an account first.")
        return

    amount = float(input("\nEnter amount to withdraw: "))

    if amount <= 0:
        print("Please enter a valid amount.")
    elif amount > balance:
        print("Insufficient balance.")
    else:
        balance = balance - amount
        print("Please collect your cash.")
        print("Current balance: ₹", balance)


# Function to check balance
def check_balance():
    if not account_created:
        print("\nPlease create an account first.")
        return

    print("\n----- ACCOUNT BALANCE -----")
    print("Account Holder:", account_name)
    print("Account Number:", account_number)
    print("Available Balance: ₹", balance)


# Function to display account details
def account_details():
    if not account_created:
        print("\nPlease create an account first.")
        return

    print("\n----- ACCOUNT DETAILS -----")
    print("Account Holder :", account_name)
    print("Account Number :", account_number)
    print("Balance        : ₹", balance)


# Main menu
while True:

    print("\n===================================")
    print("             MAIN MENU")
    print("===================================")
    print("1. Create Account")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Check Balance")
    print("5. Account Details")
    print("6. Exit")
    print("===================================")

    choice = input("Enter your choice (1-6): ")

    if choice == "1":
        create_account()

    elif choice == "2":
        deposit_money()

    elif choice == "3":
        withdraw_money()

    elif choice == "4":
        check_balance()

    elif choice == "5":
        account_details()

    elif choice == "6":
        print("\nThank you for using the Bank Management System!")
        print("Goodbye!")
        break

    else:
        print("\nInvalid choice. Please select 1 to 6.")
```
