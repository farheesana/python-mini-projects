```python
print("===== SIMPLE QUIZ GAME =====")

score = 0

answer = input("1. What is the capital of India? ").lower()

if answer == "new delhi":
    print("Correct!")
    score = score + 1
else:
    print("Wrong! The answer is New Delhi.")

answer = input("2. Which language are we learning? ").lower()

if answer == "python":
    print("Correct!")
    score = score + 1
else:
    print("Wrong! The answer is Python.")

answer = input("3. How many days are there in a week? ")

if answer == "7":
    print("Correct!")
    score = score + 1
else:
    print("Wrong! The answer is 7.")

answer = input("4. What is 5 + 5? ")

if answer == "10":
    print("Correct!")
    score = score + 1
else:
    print("Wrong! The answer is 10.")

answer = input("5. Which planet do we live on? ").lower()

if answer == "earth":
    print("Correct!")
    score = score + 1
else:
    print("Wrong! The answer is Earth.")

print("\n===== QUIZ COMPLETED =====")
print("Your score is:", score, "/ 5")
```
