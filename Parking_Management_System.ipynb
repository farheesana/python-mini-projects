````python
# ==========================================
#       PARKING MANAGEMENT SYSTEM
# ==========================================

TOTAL_SLOTS = 5

parking = {}

# Function to park a vehicle
def park_vehicle():
    if len(parking) >= TOTAL_SLOTS:
        print("\nParking is full!")
        return

    vehicle_number = input("Enter vehicle number: ")

    if vehicle_number in parking:
        print("\nVehicle is already parked.")
        return

    # Find the first available slot
    for slot in range(1, TOTAL_SLOTS + 1):
        if slot not in parking.values():
            parking[vehicle_number] = slot
            print("\nVehicle parked successfully!")
            print("Vehicle Number:", vehicle_number)
            print("Parking Slot:", slot)
            return


# Function to view parked vehicles
def view_vehicles():
    if len(parking) == 0:
        print("\nNo vehicles are currently parked.")
        return

    print("\n========== PARKED VEHICLES ==========")

    for vehicle, slot in parking.items():
        print("Vehicle Number:", vehicle)
        print("Parking Slot  :", slot)
        print("------------------------------------")


# Function to remove a vehicle
def remove_vehicle():
    vehicle_number = input("Enter vehicle number to remove: ")

    if vehicle_number in parking:
        slot = parking[vehicle_number]
        del parking[vehicle_number]

        print("\nVehicle removed successfully!")
        print("Vehicle Number:", vehicle_number)
        print("Slot", slot, "is now available.")

    else:
        print("\nVehicle not found.")


# Function to display available slots
def available_slots():
    available = TOTAL_SLOTS - len(parking)

    print("\nTotal Parking Slots     :", TOTAL_SLOTS)
    print("Occupied Parking Slots  :", len(parking))
    print("Available Parking Slots :", available)


# Main program
while True:

    print("\n======================================")
    print("       PARKING MANAGEMENT SYSTEM")
    print("======================================")

    print("1. Park Vehicle")
    print("2. View Parked Vehicles")
    print("3. Remove Vehicle")
    print("4. Check Available Slots")
    print("5. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        park_vehicle()

    elif choice == "2":
        view_vehicles()

    elif choice == "3":
        remove_vehicle()

    elif choice == "4":
        available_slots()

    elif choice == "5":
        print("\nThank you for using Parking Management System!")
        break

    else:
        print("\nInvalid choice. Please try again.")
````

# 🚗 Parking Management System

## Project Description

The **Parking Management System** is a beginner-friendly Python mini project that simulates a simple vehicle parking system.

The system allows users to park vehicles, view parked vehicles, remove vehicles, and check the number of available parking spaces.

## 🎯 Objectives

* To manage vehicle parking spaces.
* To store vehicle information.
* To assign parking slots to vehicles.
* To remove vehicles when they leave.
* To display currently parked vehicles.
* To check available parking spaces.

## ✨ Features

1. **Park Vehicle** – Parks a vehicle in an available slot.
2. **View Parked Vehicles** – Displays all currently parked vehicles.
3. **Remove Vehicle** – Removes a vehicle from the parking area.
4. **Available Slots** – Displays the number of free parking slots.
5. **Exit** – Closes the parking management system.

Parking areas need to keep track of vehicles entering and leaving.

For example, when a car enters a parking area, it is assigned a parking slot. When the car leaves, that slot becomes available again.

This project demonstrates the basic logic behind such a parking system.
