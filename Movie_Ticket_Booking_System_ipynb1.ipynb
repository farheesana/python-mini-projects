```python
# =====================================================
#          MOVIE TICKET BOOKING SYSTEM
# =====================================================

print("=" * 50)
print("       🎬 MOVIE TICKET BOOKING SYSTEM 🎬")
print("=" * 50)

# Movie details
movies = {
    1: {"name": "Avengers", "price": 200},
    2: {"name": "Inception", "price": 180},
    3: {"name": "Interstellar", "price": 220},
    4: {"name": "The Lion King", "price": 150}
}

booking = None


# -----------------------------------------------------
# Function to display movies
# -----------------------------------------------------

def show_movies():

    print("\n========== AVAILABLE MOVIES ==========")

    for number, movie in movies.items():
        print(
            number,
            ".",
            movie["name"],
            "- ₹",
            movie["price"],
            "per ticket"
        )


# -----------------------------------------------------
# Function to book tickets
# -----------------------------------------------------

def book_ticket():

    global booking

    if booking is not None:
        print("\nYou already have a booking.")
        print("Cancel the existing booking before making a new one.")
        return

    show_movies()

    try:
        movie_choice = int(input("\nSelect movie number: "))

        if movie_choice not in movies:
            print("Invalid movie choice.")
            return

        movie = movies[movie_choice]

        customer_name = input("Enter your name: ")

        seats = int(input("Enter number of tickets: "))

        if seats <= 0:
            print("Number of tickets must be greater than zero.")
            return

        total = movie["price"] * seats

        booking = {
            "name": customer_name,
            "movie": movie["name"],
            "seats": seats,
            "price": movie["price"],
            "total": total
        }

        print("\n===================================")
        print("       🎟️ BOOKING SUCCESSFUL")
        print("===================================")

        print("Customer Name :", customer_name)
        print("Movie         :", movie["name"])
        print("Tickets       :", seats)
        print("Ticket Price  : ₹", movie["price"])
        print("Total Amount  : ₹", total)

        print("===================================")

    except ValueError:
        print("Please enter a valid number.")


# -----------------------------------------------------
# Function to view booking
# -----------------------------------------------------

def view_booking():

    if booking is None:
        print("\nNo booking found.")
        return

    print("\n========== BOOKING DETAILS ==========")

    print("Customer Name :", booking["name"])
    print("Movie         :", booking["movie"])
    print("Tickets       :", booking["seats"])
    print("Ticket Price  : ₹", booking["price"])
    print("Total Amount  : ₹", booking["total"])


# -----------------------------------------------------
# Function to cancel booking
# -----------------------------------------------------

def cancel_booking():

    global booking

    if booking is None:
        print("\nNo booking available to cancel.")
        return

    print("\nBooking cancelled successfully!")
    print("Refund Amount: ₹", booking["total"])

    booking = None


# -----------------------------------------------------
# Main Menu
# -----------------------------------------------------

while True:

    print("\n")
    print("=" * 50)
    print("              MAIN MENU")
    print("=" * 50)

    print("1. 🎥 View Movies")
    print("2. 🎟️ Book Ticket")
    print("3. 📋 View Booking")
    print("4. ❌ Cancel Booking")
    print("5. 🚪 Exit")

    print("=" * 50)

    choice = input("Enter your choice (1-5): ")

    if choice == "1":

        show_movies()

    elif choice == "2":

        book_ticket()

    elif choice == "3":

        view_booking()

    elif choice == "4":

        cancel_booking()

    elif choice == "5":

        print("\nThank you for using Movie Ticket Booking System! 🎬")
        print("Have a great day! 😊")
        break

    else:

        print("\nInvalid choice. Please select 1 to 5.")
```
