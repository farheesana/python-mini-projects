````python
# ==========================================
#       LIBRARY MANAGEMENT SYSTEM
# ==========================================

books = []

# Function to add a book
def add_book():
    book_id = input("Enter Book ID: ")
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "status": "Available"
    }

    books.append(book)

    print("\nBook added successfully!")


# Function to view books
def view_books():
    if len(books) == 0:
        print("\nNo books available.")
        return

    print("\n========== BOOK LIST ==========")

    for book in books:
        print("Book ID :", book["id"])
        print("Title   :", book["title"])
        print("Author  :", book["author"])
        print("Status  :", book["status"])
        print("------------------------------")


# Function to search a book
def search_book():
    search = input("Enter book title to search: ")

    found = False

    for book in books:
        if search.lower() in book["title"].lower():
            print("\nBook Found!")
            print("Book ID :", book["id"])
            print("Title   :", book["title"])
            print("Author  :", book["author"])
            print("Status  :", book["status"])
            found = True

    if not found:
        print("\nBook not found.")


# Function to issue a book
def issue_book():
    book_id = input("Enter Book ID to issue: ")

    for book in books:
        if book["id"] == book_id:

            if book["status"] == "Available":
                book["status"] = "Issued"
                print("\nBook issued successfully!")
            else:
                print("\nBook is already issued.")

            return

    print("\nBook ID not found.")


# Function to return a book
def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in books:
        if book["id"] == book_id:

            if book["status"] == "Issued":
                book["status"] = "Available"
                print("\nBook returned successfully!")
            else:
                print("\nThis book was not issued.")

            return

    print("\nBook ID not found.")


# Main program
while True:

    print("\n===================================")
    print("      LIBRARY MANAGEMENT SYSTEM")
    print("===================================")

    print("1. Add Book")
    print("2. View Books")
    print("3. Search Book")
    print("4. Issue Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        view_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        issue_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("\nThank you for using Library Management System!")
        break

    else:
        print("\nInvalid choice. Please try again.")
````

# 📚 Library Management System

## Project Description

The **Library Management System** is a beginner-friendly Python mini project designed to manage books in a library.

The program allows the user to add books, view available books, search for books, issue books, and return books.

This project demonstrates how Python can be used to develop a simple real-world management system.

## 🎯 Objectives

* To maintain a list of books in a library.
* To add new books to the library.
* To search for a particular book.
* To issue books to users.
* To return issued books.
* To display the current status of books.

## ✨ Features

1. **Add Book** – Add a new book with Book ID, title, and author.
2. **View Books** – Display all books and their current status.
3. **Search Book** – Search for a book using its title.
4. **Issue Book** – Change the book status from Available to Issued.
5. **Return Book** – Change the book status from Issued back to Available.
6. **Exit** – Close the Library Management System.

A real library maintains information about books, including:

* Book ID

For example, when a student borrows a book, its status changes from **Available** to **Issued**. When the student returns it, the status changes back to **Available**.
