```python
# ==========================================
#          PALINDROME CHECKER
# ==========================================

print("===================================")
print("       PALINDROME CHECKER")
print("===================================")

text = input("Enter a word, number, or sentence: ")

# Convert to lowercase and remove spaces
cleaned_text = text.lower().replace(" ", "")

# Reverse the text
reversed_text = cleaned_text[::-1]

print("\nOriginal input:", text)
print("Reversed input:", reversed_text)

# Check palindrome
if cleaned_text == reversed_text:
    print("Result: It is a Palindrome!")
else:
    print("Result: It is NOT a Palindrome!")

print("\nThank you for using Palindrome Checker!")
```

# Palindrome Checker

## Project Overview
The Palindrome Checker is a Python mini-project that checks
whether a word, number, or sentence reads the same forward
and backward.

## Objective
To check whether the given input is a palindrome using Python.

## Python Concepts Used
- Variables
- Strings
- Input and Output
- String slicing
- If-else statements

## Examples
- madam → Palindrome
- 121 → Palindrome
- level → Palindrome
- hello → Not a Palindrome
